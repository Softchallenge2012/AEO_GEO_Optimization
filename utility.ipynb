```python
import json
import pandas as pd
import re
```

```python
df = pd.read_json('tmobile_detailed_specs.json')
df.head()
```

```python
df['features'] = df['specs'].apply(lambda t: re.split(r'\bQuick Facts\b', str(t['section_77']))[0] )
df['product'] = df['url'].apply(lambda t: t.split('/')[-1])
df.head()
```

```python
df[['product','features']].to_csv('tmobile_detailed_specs.csv',index=False)
```
